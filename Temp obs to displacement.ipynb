```python
file_name1 = r'C:\Users\21924032\GitHub\Yutao_PhD_codes\Aim2 Mode decomposition\notebooks\V3.0\PL3 Stacked data_mode_3 - density fit - single_tanh\stacked_LV3_NWSROW-2204.nc'
stacked_lv3 = xr.open_dataset(file_name1).copy(deep=True)
stacked_lv3['rhofit'].plot(y='z_interp', x='time')
```

```python
import matplotlib.pyplot as plt
import cmocean

fig, axes = plt.subplots(2, 2, figsize=(16, 10),)
axes = axes.flatten()
# --- Color map plots ---
stacked_lv3['TEMP'].plot(ax=axes[0], cmap='cmo.thermal')
axes[0].set_title('Temperature')
axes[0].set_ylabel('Depth (m)')
stacked_lv3['sea_water_potential_density'].plot(ax=axes[1], cmap='cmo.dense')
axes[1].set_title('Density')
axes[1].set_ylabel('Depth (m)')
# --- Hide unused subplot (optional) ---
axes[2].axis('off')

# --- Line plot for A_n modes ---
line_styles = ['-', '--', ':']
colors = ['tab:blue', 'tab:orange', 'tab:green']

# Assuming stacked_lv3['A_n'] has a 'mode' dimension
for n, (ls, color) in enumerate(zip(line_styles, colors)):
    stacked_lv3['A_n'].sel(modes=n).plot(ax=axes[3], linestyle=ls, color=color, label=f'Mode={n+1}')

axes[3].set_title('A_n by Mode')
axes[3].legend()
axes[3].set_ylabel('Displacement (m)')  # adjust if needed

for ax in axes:
    # ax.xaxis.set_major_locator(plt.MaxNLocator(6))  # show ~6 ticks
    ax.xaxis.set_major_formatter(plt.matplotlib.dates.DateFormatter('%Y-%m-%d'))
    ax.set_xlabel('')
    ax.set_title('')
fig.autofmt_xdate()  # auto rotate date labels    
plt.tight_layout()
plt.show()
```
